```python
MY_IMAGE = "/1.jpeg"
img = cv2.imread(MY_IMAGE)
img.shape
```

```python
# Step 4: display the image
# IMPORTANT: OpenCV stores colours in B, G, R order but Matplotlib expects R, G, B order.
# So we must convert first, otherwise the colours look wrong.

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(img)                       # without conversion
plt.title("Wrong colours (BGR)")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(img_rgb)                   # with conversion
plt.title("Correct colours (RGB)")
plt.axis("off")

plt.show()
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

print("Colour Image shape:", img.shape)
print("Gray image shape:", gray.shape)

plt.figure(figsize=(5, 5))
plt.imshow(gray, cmap="gray")
plt.axis("off")
plt.show()
```

```python
height= gray.shape[0]
width=gray.shape[1]

new_img_size=cv2.resize(gray,(width//2,height//2))

print("Original size",gray.shape)
print("New size",new_img_size.shape)

plt.figure(figsize=(10,4))
plt.imshow(new_img_size,cmap="gray")
plt.title("Image en niveaux de gris")
plt.axis("off")
plt.show()
```

```python
brighter = cv2.convertScaleAbs(gray, alpha= 1, beta= 60)
darker = cv2. convertScaleAbs(gray, alpha= 1, beta = -60)

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(darker, cmap="gray", vmin=0, vmax=255)
plt.title("Darker (beta = 60)")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(gray, cmap="gray", vmin=0, vmax=255)
plt.title("Original")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(brighter, cmap="gray", vmin=0, vmax=255)
plt.title("Brighter (beta = +60)")
plt.axis("off")
plt.show()
```

```python
high_contrast = cv2.convertScaleAbs(gray, alpha= 1.8, beta= 0)
low_contrast = cv2.convertScaleAbs(gray, alpha= 0.5, beta= 0)

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(low_contrast, cmap="gray", vmin=0, vmax=255)
plt.title("low_contrast (alpha = 0.5)")
plt.axis("off")

plt.subplot(1, 3, 1)
plt.imshow(gray, cmap="gray", vmin=0, vmax=255)
plt.title("Original")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(high_contrast, cmap="gray", vmin=0, vmax=255)
plt.title("High contrast (alpha = 1.8)")
plt.axis("off")

plt.show()
```
