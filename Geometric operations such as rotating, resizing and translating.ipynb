```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
MY_IMAGE = "/content/Mount-Everest.webp"
```

```python
img = cv2.imread(MY_IMAGE)
height, width = img.shape[:2]
window_name = "2D Transformations"
```

```python
# Display the original and transformed images
plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')
```

```python
# 1. Translate the image
translation_matrix = np.float32([[1, 0, 50], [0, 1, 30]]) # translate by 50px horizontally, 30px vertically
img_translated = cv2.warpAffine(img, translation_matrix, (width, height))

plt.subplot(2, 2, 2)
plt.imshow(cv2.cvtColor(img_translated, cv2.COLOR_BGR2RGB))
plt.title('Translated Image')
plt.axis('off')
```

```python
# 2. Resize the image to 50% of its original size
img_resized = cv2.resize(img, (int(width * 0.5), int(height * 0.5)))

plt.subplot(2, 2, 3)
plt.imshow(cv2.cvtColor(img_resized, cv2.COLOR_BGR2RGB))
plt.title('Resized Image')
plt.axis('off')
```

```python

# 3. Rotate the image by 60 degrees about its center
center = (width // 2, height // 2)
rotation_matrix = cv2.getRotationMatrix2D(center, 60, 1.0) # 60 degrees, scale = 1.0
img_rotated = cv2.warpAffine(img, rotation_matrix, (width, height))

plt.subplot(2, 2, 4)
plt.imshow(cv2.cvtColor(img_rotated, cv2.COLOR_BGR2RGB))
plt.title('Rotated Image')
plt.axis('off')

plt.tight_layout()
plt.show()
```
