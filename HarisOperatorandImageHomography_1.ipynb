```python
#Writing code to use Haris Operator to detect corners in an empty room- by Akshat

from google.colab.patches  import cv2_imshow
from google.colab import files
import cv2
import numpy as np
img_HarisOperator = cv2.imread("/content/HARRIS_OPERATOR_IMAGE.webp")
print("original image")
cv2_imshow(img_HarisOperator)
gray = cv2.cvtColor(img_HarisOperator,cv2.COLOR_BGR2GRAY)
gray = np.float32(gray)
corners = cv2.cornerHarris(
    gray,
    blockSize=2,
    ksize=3,
    k=0.04
)

result = img_HarisOperator.copy()
result[corners > 0.0001 * corners.max()] = [0, 0, 255]

cv2_imshow(result)
```

```python
#Doing Image Homography - By Akshat
#CAN ONLY work on flat surface images like documents

from google.colab.patches  import cv2_imshow
from google.colab import files
import cv2
import numpy as np
img_homography = cv2.imread("/content/ImageHomography_1.jpeg")
print("original image")
cv2_imshow(img_homography)

points1 = np.float32([[118,78],[520,128],[471,399],[74,331]])
points2 = np.float32([[0,0],[400,0],[400,300],[0,300]])
H, status = cv2.findHomography(points1,points2)

print("Homography Matrix")
print(H)

result = cv2.warpPerspective(img_homography,H,(400,300))
result_rgb = cv2.cvtColor(result, cv2.COLOR_BGR2RGB)
print("Result")
cv2_imshow(result_rgb)
```
