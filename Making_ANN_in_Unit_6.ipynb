```python
from sklearn.datasets import load_iris #Sk learn is used to train ml models, to test , split etc
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential #Tensorflow is used to make neural networks and seqential means all the models are sequential
from tensorflow.keras.layers import Dense
from sklearn.metrics import accuracy_score

d = load_iris()
x = d.data
y = d.target
x_train, x_test, y_train, y_test = train_test_split(x, y)

model = Sequential(
    [
        Dense(10, activation='relu', input_shape=(4,)),
        Dense(8, activation='relu'), #Relu is always used in hidden layers
        Dense(3, activation='softmax') #multiclass classification softmax will be used for that
    ]
)

model.compile(
    loss='sparse_categorical_crossentropy', optimizer='adam', metrics=['accuracy']
) #compiling the model

model.fit(x_train, y_train, epochs=50) #Adam optimizes the weights of the model. Higher Epochs's value, then it is underfitting, higher epochs will lead to memorization of training data.

loss, Accuracy = model.evaluate(x_test, y_test)
print("Loss: ", loss)
print("Accuracy: ", Accuracy)
```
